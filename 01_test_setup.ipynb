```python
def same_molecule(smiles1, smiles2):
    if pd.isna(smiles1) or pd.isna(smiles2):
        return False
    
    mol1 = Chem.MolFromSmiles(smiles1)
    mol2 = Chem.MolFromSmiles(smiles2)
    
    if mol1 is None or mol2 is None:
        return False
    
    return Chem.MolToInchiKey(mol1) == Chem.MolToInchiKey(mol2)
```

```python
same_molecule(
    df_model.loc[10865, "products 1"],
    df_model.loc[10865, "reactants 2"]
)
```

```python
def product_matches_reactant(row):
    product = row["products 1"]
    
    if pd.isna(product):
        return False
    
    for col in ["reactants 1", "reactants 2", "reactants 3", "reactants 4", "reactants 5"]:
        if same_molecule(product, row[col]):
            return True
    
    return False
```

```python
def make_reactant_fingerprint(row):
    mol1 = Chem.MolFromSmiles(row["reactants 1"])
    mol2 = Chem.MolFromSmiles(row["reactants 2"])

    fp1 = morgan_gen.GetFingerprint(mol1)
    fp2 = morgan_gen.GetFingerprint(mol2)

    fp1 = list(fp1)
    fp2 = list(fp2)

    if fp1 > fp2:
        fp1, fp2 = fp2, fp1

    return np.array(fp1 + fp2, dtype=np.uint8)
```

```python
test_fp = make_reactant_fingerprint(df_ml.iloc[0])

print(test_fp.shape)
print(test_fp.dtype)
print(test_fp[:20])
```

```python
X = np.zeros((len(df_ml), 4096), dtype=np.uint8)

for i, (_, row) in enumerate(df_ml.iterrows()):
    X[i] = make_reactant_fingerprint(row)

print(X.shape)
print(X.dtype)
```

```python
from rdkit import Chem

def same_molecule(smiles1, smiles2):
    if pd.isna(smiles1) or pd.isna(smiles2):
        return False

    mol1 = Chem.MolFromSmiles(smiles1)
    mol2 = Chem.MolFromSmiles(smiles2)

    if mol1 is None or mol2 is None:
        return False

    return Chem.MolToInchiKey(mol1) == Chem.MolToInchiKey(mol2)


def product_matches_reactant(row):
    product = row["products 1"]

    if pd.isna(product):
        return False

    for col in [
        "reactants 1",
        "reactants 2",
        "reactants 3",
        "reactants 4",
        "reactants 5"
    ]:
        if same_molecule(product, row[col]):
            return True

    return False


df_model["product_matches_reactant"] = df_model.apply(
    product_matches_reactant,
    axis=1
)

print(df_model["product_matches_reactant"].value_counts())
```

```python
from collections import Counter
from sklearn.preprocessing import MultiLabelBinarizer

def make_multihot(bags, train_idx, min_count, prefix):
    counts = Counter(
        item
        for i in train_idx
        for item in bags.iloc[i]
    )

    vocab = [
        item for item, count in counts.items()
        if count >= min_count
    ]

    mlb = MultiLabelBinarizer(classes=vocab)
    mlb.fit([vocab])

    X = mlb.transform(
        bags.apply(
            lambda s: [item for item in s if item in vocab]
        )
    )

    names = [prefix + item for item in vocab]

    return X.astype(np.uint8), names


# Components = reagents + catalysts
X_components, component_names = make_multihot(
    comp_bag,
    random_train_idx,
    min_count=100,
    prefix="comp_"
)

# Solvents
X_solvents, solvent_names = make_multihot(
    solv_bag,
    random_train_idx,
    min_count=50,
    prefix="solv_"
)

# Numeric conditions
condition_numeric = np.column_stack([
    df_ml["temperature /C"].to_numpy(dtype=np.float32),
    np.log1p(df_ml["time /h"].to_numpy(dtype=np.float32)),
    comp_bag.apply(len).to_numpy(dtype=np.float32),
    solv_bag.apply(len).to_numpy(dtype=np.float32)
])

# Combine everything
X_conditions = np.hstack([
    condition_numeric,
    X_components,
    X_solvents
])

condition_names = [
    "temperature",
    "log_time",
    "n_components",
    "n_solvents"
] + component_names + solvent_names

print("Condition matrix shape:", X_conditions.shape)
print("Number of condition features:", len(condition_names))
print("\nFirst 30 features:")
print(condition_names[:30])
```

```python
from rdkit import Chem
import numpy as np, pandas as pd

smi1 = df_ml["reactants 1"].astype(str).values
smi2 = df_ml["reactants 2"].astype(str).values

alkene = Chem.MolFromSmarts("[CX3;!a]=[CX3;!a]")
extra_e = [Chem.MolFromSmarts(s) for s in [
    "[c,$(C=C)][Cl,Br,I]",
    "[c,$(C=C)]OS(=O)(=O)[#6]",
    "[c,$(C=C)]B(O)O",
    "[c,$(C=C)]B1OCCO1",
    "c[N+]#N",
    "[CX4][Cl,Br,I]",
    "[CX3](=O)[Cl,Br,I]",
]]

def has_electrophile(smi):
    m = Chem.MolFromSmiles(str(smi))
    return m is not None and any(m.HasSubstructMatch(p) for p in extra_e)

def has_alkene(smi):
    m = Chem.MolFromSmiles(str(smi))
    return m is not None and m.HasSubstructMatch(alkene)

e1 = df_ml["reactants 1"].apply(has_electrophile).values
e2 = df_ml["reactants 2"].apply(has_electrophile).values
a1 = df_ml["reactants 1"].apply(has_alkene).values
a2 = df_ml["reactants 2"].apply(has_alkene).values

side2 = np.where(e1 & ~e2, False,
        np.where(e2 & ~e1, True,
        np.where(~a1 & a2, False,
        np.where(a1 & ~a2, True, False))))
tier = np.where(e1 ^ e2, 1, np.where(a1 ^ a2, 2, 3))

groups_halide = np.where(side2, smi2, smi1)
print("tier counts:", pd.Series(tier).value_counts().sort_index().to_dict())
print("unique groups:", len(set(groups_halide)))
print("largest:", pd.Series(groups_halide).value_counts().head(5).tolist())
```
